给你一个长度为 n 的整数数组 nums 和 一个目标值 target。请你从 nums 中选出三个整数，使它们的和与 target 最接近。

返回这三个数的和。

假定每组输入只存在恰好一个解。

链接：https://leetcode.cn/problems/3sum-closest

示例 1：

输入：nums = [-1,2,1,-4], target = 1
输出：2
解释：与 target 最接近的和是 2 (-1 + 2 + 1 = 2) 。

示例 2：

输入：nums = [0,0,0], target = 1
输出：0

提示：

3 <= nums.length <= 1000
-1000 <= nums[i] <= 1000
-104 <= target <= 104

思路：list排序，以长度为3的移动滑动窗口分别计算滑动窗口的求和值，如果求和值==targets，找到目标。否则，记录下距离最小对应的滑动窗口，如果求和值比target大，那么置换滑动窗口左边的数，选择一个距离比当前小的数对应的list，否则置换滑动窗口右边的数，选择滑动窗口右边的数，选择一个距离比当前小的list。

上面这个思路的错误在于，只调整最左边和最右边只是让总和变小，但是不能确定就是最优值，因为按照这个思路，当求和值>target时，调整中间值或者右边的值也可以使得总和变小，思路错误

```python
from typing import List
class Solution:
    def threeSumClosest(self, nums: List[int], target: int) -> int:
        assert len(nums)>=3 and len(nums)<=1000
        assert target>=-104 and target<=104
        for _ in nums:
            if _<-1000 or _>1000: #确认数据的边界值
                return False 
        nums.sort()
        dis=9999999 #计录最小距离
        sum_num=0 #计算最小的和
        l_=[]
        n=len(nums)
        l,r=0,0 #记录起始位置
        for i in range(n-2):
            c_l=nums[i:i+3]
            num=sum(c_l)
            if abs(num-target)<dis:
                dis=abs(num-target)
                l_=c_l
                sum_num=num
                l,r=i,i+2
            else:# 距离不变或者比距离变大不做操作
                pass 
        #print(l_)
        if sum_num>target: #向左寻找
            for i in range(l):
                _=nums[i]+l_[1]+l_[2]
                if abs(_-target)<dis:
                    l_[0]=nums[i]
                    dis=abs(_-target)
        elif sum_num==target:#找到最小值
            return target
        else: #向右寻找
            for i in range(r,n):
                _=l_[0]+l_[1]+nums[i]
                if abs(_-target)<dis:
                    l_[2]=nums[i]
                    dis=abs(_-target)
        return sum(l_)
            
```

```python
solution=Solution()
print(solution.threeSumClosest([-1,2,1,-4],1))
print(solution.threeSumClosest( [0,0,0],1))
print(solution.threeSumClosest( [1,2,4,8,16,32,64,128],82)) #
```
