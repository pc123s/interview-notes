'''
给你一个字符串 croakOfFrogs，它表示不同青蛙发出的蛙鸣声（字符串 "croak" ）的组合。由于同一时间可以有多只青蛙呱呱作响，所以 croakOfFrogs 中会混合多个 “croak” 。

请你返回模拟字符串中所有蛙鸣所需不同青蛙的最少数目。

要想发出蛙鸣 "croak"，青蛙必须 依序 输出 ‘c’, ’r’, ’o’, ’a’, ’k’ 这 5 个字母。如果没有输出全部五个字母，那么它就不会发出声音。如果字符串 croakOfFrogs 不是由若干有效的 "croak" 字符混合而成，请返回 -1 。


输入：croakOfFrogs = "croakcroak"
输出：1 
解释：一只青蛙 “呱呱” 两次

输入：croakOfFrogs = "crcoakroak"
输出：2 
解释：最少需要两只青蛙，“呱呱” 声用黑体标注
第一只青蛙 "crcoakroak"
第二只青蛙 "crcoakroak"

输入：croakOfFrogs = "croakcrook"
输出：-1
解释：给出的字符串不是 "croak" 的有效组合。
字符串长度1-10000

'''

'''
我的思路，将字符串看出一个排出的字符流，利用容器去接这些字符流，每个容器里面只能装固定的5个字符‘c’, ’r’, ’o’, ’a’, ’k’，分别编号1-2-3-4-5，一个容器装满马上释放，如果一个容器没有装满，出现新的字符那么开始构建第二个容器，依次类推，直到字符串遍历结束。计算使用了多少个容器
'''

```python
#这种方法可以求解但是时间复杂度过高
class Solution:
    def minNumberOfFrogs(self, croakOfFrogs: str) -> int:
        res = []
        dic = {'c': 'r', 'r': 'o', 'o': 'a', 'a': 'k', 'k': ''}  # 查看当前字符的下一个字符是什么
        for _ in croakOfFrogs:
            #print(_)
            if _ == 'c':  # 遇到c判断其中是否有容器
                if len(res) == 0:  # 没有创建一个
                    arr = []
                    arr.append('c')
                    res.append(arr)
                    continue
                else:
                    triger=False #判断其中是否有容器中是否有完整的蛙声序列，是 True, 否 False
                    for i, arr_ in enumerate(res):
                        if arr_[-1] == 'k': #容器末端以k结尾继续装载
                            triger=True
                            res[i].append('c')
                            break
                        else:
                            pass
                    if triger==False:#说明里面没有完整的蛙声容器
                            arr = []
                            arr.append('c')
                            res.append(arr)
            else:
                triger=False
                for i, arr_ in enumerate(res):
                    if dic[arr_[-1]] == _:  # 当前字符刚好是容器需要的那个字符
                        arr_.append(_)
                        res[i] = arr_
                        triger=True
                        break
                if triger==False:#说明数据没有找到
                    return -1
        #print(res)
        res_ = [1 for arr in res if len(arr) % 5 == 0]
        if sum(res_)!=len(res):
            return -1
        else:
            return len(res_)
```

```python
solution=Solution()
print(solution.minNumberOfFrogs('croakcroak'))
print(solution.minNumberOfFrogs('crcoakroak'))
print(solution.minNumberOfFrogs('croakcrook'))
print(solution.minNumberOfFrogs('okcracakroroakacckor'))
print(solution.minNumberOfFrogs("croakcroa"))
```

方法2：记录同时有多少青蛙在叫即可，遍历娃叫的字符串，娃叫以c开始，k结束，听到娃叫记录一只青蛙，青蛙叫声没有结束时，每听到一声娃叫，青蛙数目+1，叫声结束后数目减1，我们需要在记录在整个蛙鸣过程中，青蛙同时叫声的最大值。如果最后有叫声没有以k结束，说明娃叫不完整返回-1，在记录字符串的同时，类似于消消乐，一定要保证左边方块大于右边才行。（官方解法）

```python
class Solution:
    def minNumberOfFrogs(self, croakOfFrogs: str) -> int:
        dic={'c':0,'r':0,'o':0,'a':0,'k':0}
        for _ in croakOfFrogs:
            dic[_]+=1
        if not dic['c']==dic['r']==dic['o']==dic['a']==dic['k']: #无法凑齐娃叫，-1
            return -1
        
        now=0 #记录同时有多少娃叫
        res=0 #记录最后的结果
        dic={'c':0,'r':0,'o':0,'a':0,'k':0}
        for _ in croakOfFrogs:
            dic[_]+=1
            
            if dic['c']>=dic['r']>=dic['o']>=dic['a']>=dic['k']: #确保左边的始终大于右边
                if _=="c":
                    dic["c"]+=1
                    now+=1
                    res=max(res,now)
                elif _=='r':
                    dic['r']+=1
                elif _=='o':
                    dic['o']+=1
                elif _=='a':
                    dic['a']+=1
                else:
                    dic['k']+=1
                    now-=1
            else:#出现右边大于左边的情况
                return -1 
        return res
```

```python
solution=Solution()
print(solution.minNumberOfFrogs("croakcroak"))
print(solution.minNumberOfFrogs('crcoakroak'))
print(solution.minNumberOfFrogs('croakcrook'))
print(solution.minNumberOfFrogs('okcracakroroakacckor'))
print(solution.minNumberOfFrogs("crocakcroraoakk"))
```

方法3：更通俗的解法，消消乐，集合5个进行消除，消除时看看场上还有多少个未消除的方块-有多少个c（说明这些娃叫还没结束）。记录每个消除时娃叫的最大值，即方块消除时多少个c

```python
class Solution:
    def minNumberOfFrogs(self, croakOfFrogs: str) -> int:
        dic={'c':0,'r':0,'o':0,'a':0,'k':0}
        for _ in croakOfFrogs:
            dic[_]+=1
        if not dic['c']==dic['r']==dic['o']==dic['a']==dic['k']: #无法凑齐娃叫，-1
            return -1        
        dic={'c':0,'r':0,'o':0,'a':0,'k':0}
        res=0# 记录消除时场上最多的方块数
        for _ in croakOfFrogs:
            dic[_]+=1
            
            if dic['c']>=dic['r']>=dic['o']>=dic['a']>=dic['k']: #确保左边的始终大于右边
                if dic['c']>=dic['r']>=dic['o']>=dic['a']>=dic['k']>=1:#满足消消乐条件
                    res=max(dic['c'],res) #记录当时方块的最大值
                    for c in 'croak': #消消乐一次
                        dic[c]-=1
            else:#出现右边大于左边的情况
                return -1 
        return res
```

```python
solution=Solution()
print(solution.minNumberOfFrogs("croakcroak"))
print(solution.minNumberOfFrogs('crcoakroak'))
print(solution.minNumberOfFrogs('croakcrook'))
print(solution.minNumberOfFrogs('okcracakroroakacckor'))
print(solution.minNumberOfFrogs("crocakcroraoakk"))
```
